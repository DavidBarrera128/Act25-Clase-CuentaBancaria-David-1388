## Cuenta Bancara


![imagen](infoclassxxx.png.png)

```python
class CuentaBancaria:
    # Atributos de la clase
    def __init__(self):
        self.titular = ""
        self.numero_cuenta = 0
        self.saldo = 0

    # Función para ingresar los datos
    def leer_datos(self):
        print("\n--- Registro de Cuenta ---")
        self.titular = input("Ingrese el nombre del titular: ")
        # Convertimos a int como solicitaste
        self.numero_cuenta = int(input("Ingrese el número de cuenta (solo números): "))
        self.saldo = int(input("Ingrese el saldo inicial (solo números): "))

    # Función para mostrar los datos
    def mostrar_datos(self):
        print("\n--- Información de la Cuenta ---")
        print(f"Titular: {self.titular}")
        print(f"Número de Cuenta: {self.numero_cuenta}")
        print(f"Saldo Actual: ${self.saldo}")

# --- Creación de Objetos y Ejemplos ---

# 1. Creamos el objeto cuenta1
cuenta1 = CuentaBancaria()
print("Ejemplo para cuenta 1: (Titular: Juan Perez, Cuenta: 12345, Saldo: 5000)")
cuenta1.leer_datos()

# 2. Creamos el objeto cuenta2
cuenta2 = CuentaBancaria()
print("\nEjemplo para cuenta 2: (Titular: Maria Lopez, Cuenta: 67890, Saldo: 10000)")
cuenta2.leer_datos()

# Mostramos los resultados finales
print("\n--- RESULTADOS FINALES ---")
cuenta1.mostrar_datos()
cuenta2.mostrar_datos()
print("Programa Realizado por David Barrera NL = 1388")
```
